# Olist Profiling Observations

This notebook confirms or contradicts the known quirks based on profiling.

### 1. `customer_id` uniqueness and `customer_unique_id`
- **Observation**: The profiling summary shows `customer_id` has 99,441 unique values in 99,441 rows (100% unique per order). `customer_unique_id` has 96,096 unique values, confirming roughly 3% of people have multiple orders.

### 2. Zip code prefixes
- **Observation**: `customer_zip_code_prefix` and `geolocation_zip_code_prefix` were ingested as strings (dtype `object`), preventing loss of leading zeros. We observe 14,994 unique zips in customers and 19,015 in geolocation.

### 3. Geolocation duplicates and bounds
- **Observation**: The `geolocation` table has 1,000,163 rows, but only 19,015 unique prefixes and 717,360 unique coordinates. This confirms massive duplication per zip prefix. The minimum latitude is -36.60 and maximum is 45.06 (with max longitude 121.10), confirming out-of-Brazil coordinates.

### 4. Timestamps are strings and null depending on status
- **Observation**: `order_delivered_customer_date` has 2,965 nulls (2.98%) and `order_delivered_carrier_date` has 1,783 nulls (1.79%), consistent with some orders not reaching delivery status.

### 5. Delivered lack timestamp / No items
- **Observation**: Null count in `order_delivered_customer_date` (2965) vs `order_status` 'delivered' (96478). We will check this closer in cleaning, but profiling shows missing timestamps. `orders` has 99,441 rows but `order_items` only has 98,666 unique `order_id`s. 99441 - 98666 = 775 orders with no items! Confirmed.

### 6. Payment rows
- **Observation**: `order_payments` has 103,886 rows for 99,440 unique `order_id`s, confirming multiple payments per order. Payment value minimum is 0.0, and `payment_type` has 'not_defined' (3 rows).

### 7. Review ID repeats
- **Observation**: `order_reviews` has 99,224 rows but only 98,410 unique `review_id`s and 98,673 unique `order_id`s. This confirms multiple reviews per order and repeating review ids.

### 8. Null categories
- **Observation**: `product_category_name` has 610 nulls (1.85% of 32,951 products). Confirmed.

### 9. Order volume over time
- **Observation**: Min purchase timestamp is in 2016 and max in late 2018. Confirmed.

### 10. Inconsistent city names
- **Observation**: There are 4,119 unique city names in `customers` but more than that in `geolocation` (8,011). Spelling inconsistencies likely inflate this number.

### 11. Price and freight right-skewed
- **Observation**: Price mean is 120.65 but median is 74.99 (skew 7.92). Freight mean is 19.99 with median 16.26 (skew 5.64). Heavily right-skewed with extreme max values (price max 6735, freight max 409). Confirmed.
