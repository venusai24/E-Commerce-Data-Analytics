# Olist Business Analysis

This notebook queries the KPI views in the data warehouse to answer the core business questions outlined in the design plan.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine
from olist_analytics.config import get_db_url

# Set up database connection
try:
    engine = create_engine(get_db_url())
except Exception as e:
    print("Database URL not configured. Ensure Postgres is running.")
    engine = None
```

### 1. Monthly Revenue Trends
Query the `kpi.monthly_revenue` view to plot revenue over time.

```python
if engine:
    df_revenue = pd.read_sql("SELECT * FROM kpi.monthly_revenue", engine)
    df_revenue['date'] = pd.to_datetime(df_revenue['year'].astype(str) + '-' + df_revenue['month'].astype(str) + '-01')
    
    plt.figure(figsize=(12, 6))
    sns.lineplot(data=df_revenue, x='date', y='total_item_revenue')
    plt.title('Monthly Item Revenue')
    plt.show()
```

### 2. Delivery Performance
Query the `kpi.delivery_performance` view to observe SLA breaches.

```python
if engine:
    df_delivery = pd.read_sql("SELECT * FROM kpi.delivery_performance", engine)
    df_delivery['date'] = pd.to_datetime(df_delivery['year'].astype(str) + '-' + df_delivery['month'].astype(str) + '-01')
    
    plt.figure(figsize=(12, 6))
    sns.lineplot(data=df_delivery, x='date', y='pct_late_deliveries')
    plt.title('Percentage of Late Deliveries by Month')
    plt.ylabel('% Late')
    plt.show()
```

### 3. Seller Performance
Query the `kpi.seller_performance` view.

```python
if engine:
    df_sellers = pd.read_sql("SELECT * FROM kpi.seller_performance ORDER BY total_revenue DESC LIMIT 10", engine)
    
    plt.figure(figsize=(10, 6))
    sns.barplot(data=df_sellers, x='total_revenue', y='seller_id')
    plt.title('Top 10 Sellers by Revenue')
    plt.show()
```
